```python
import matplotlib.pyplot as plt
import seaborn as sbn

titanic = sbn.load_dataset("titanic")
```

```python
titanic.head()
```

```python
titanic.tail()
```

```python
titanic.info()
```

```python
sbn.countplot(data=titanic,x="sex",hue="survived",palette="Set1")
plt.title("Survival Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Survived")
plt.show()
```

```python
sbn.histplot(data=titanic,x="age",color="skyblue", edgecolor="black",kde=True)

plt.title("Titanic Passenger Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
sbn.barplot(data=titanic,x="class",y="survived",palette=["red", "blue", "green"])
plt.title("Survival by Class")
plt.xlabel("Class")
plt.ylabel("Survived")
plt.show()
```
